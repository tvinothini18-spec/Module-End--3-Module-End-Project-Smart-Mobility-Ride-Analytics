##●Prepare and clean the dataset using NumPy and Pandas

```python
import pandas as pd
import numpy as np
from scipy import stats
import warnings

warnings.filterwarnings("ignore")

# Dataset URL
url = "https://github.com/GeethaGunasekaran1/Dataset_rep/raw/refs/heads/main/Trip_Analysis.xlsx"

# Read Excel File
df = pd.read_excel(url)

print("\nDataset Loaded Successfully")

print("\nFirst 5 Rows")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns.tolist())

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

duplicate_count = df.duplicated().sum()

print(f"\nDuplicate Rows Found: {duplicate_count}")

df = df.drop_duplicates().copy()

print("Duplicate Rows Removed Successfully")

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    median_value = df[col].median()

    df[col] = df[col].fillna(median_value)

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:

    if not df[col].mode().empty:

        mode_value = df[col].mode()[0]

        df[col] = df[col].fillna(mode_value)

print("\nMissing Values Handled Successfully")

for col in df.columns:

    if 'date' in col.lower() or 'time' in col.lower():

        if df[col].dtype == 'object':
            df[col] = pd.to_datetime(
                df[col],
                format='mixed',
                errors='coerce',
                dayfirst=True
            )

            print(f"{col} converted to datetime")

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)

    upper_bound = Q3 + (1.5 * IQR)

    df = df[
        (df[col] >= lower_bound) &
        (df[col] <= upper_bound)
    ]

print("\nOutliers Removed Successfully")

pickup_cols = [
    col for col in df.columns
    if 'pickup' in col.lower()
]

drop_cols = [
    col for col in df.columns
    if 'drop' in col.lower()
]

if pickup_cols and drop_cols:

    pickup_col = pickup_cols[0]

    drop_col = drop_cols[0]

    if (
        pd.api.types.is_datetime64_any_dtype(df[pickup_col]) and
        pd.api.types.is_datetime64_any_dtype(df[drop_col])
    ):

        df['Trip_Duration_Minutes'] = (
            (df[drop_col] - df[pickup_col]).dt.total_seconds() / 60
        )

        print("\nTrip Duration Feature Created")

print("\nRemaining Missing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe())

print("\nData Cleaning Completed Successfully")
```

##●Use Probability concepts (basic probability, conditional probability, Bayes’ theorem, distributions)

```python
from scipy.stats import norm, binom

url = "https://github.com/GeethaGunasekaran1/Dataset_rep/raw/refs/heads/main/Trip_Analysis.xlsx"

df = pd.read_excel(url)

print("\nDataset Loaded Successfully")

df = df.drop_duplicates().copy()

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    median_value = df[col].median()

    df[col] = df[col].fillna(median_value)

# Fill missing categorical values
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:

    if not df[col].mode().empty:

        mode_value = df[col].mode()[0]

        df[col] = df[col].fillna(mode_value)

print("\nData Cleaning Completed")

print("\n================ BASIC PROBABILITY ================")

total_rides = len(df)

print(f"\nTotal Number of Rides: {total_rides}")

# Choose first categorical column
if len(categorical_cols) > 0:

    category_col = categorical_cols[0]

    print(f"\nUsing Column: {category_col}")

    # Most frequent category
    most_common_category = df[category_col].mode()[0]

    favorable_cases = len(
        df[df[category_col] == most_common_category]
    )

    basic_probability = favorable_cases / total_rides

    print(f"\nMost Common Category: {most_common_category}")

    print(f"Probability of {most_common_category}:")
    print(basic_probability)

print("\n============ CONDITIONAL PROBABILITY =============")

if len(categorical_cols) >= 2:

    event_A_col = categorical_cols[0]

    event_B_col = categorical_cols[1]

    event_A = df[event_A_col].mode()[0]

    event_B = df[event_B_col].mode()[0]

    # P(A and B)
    intersection = len(
        df[
            (df[event_A_col] == event_A) &
            (df[event_B_col] == event_B)
        ]
    )

    # P(B)
    total_B = len(
        df[df[event_B_col] == event_B]
    )

    # Conditional Probability P(A|B)
    conditional_probability = intersection / total_B

    print(f"\nP({event_A} | {event_B}) =")
    print(conditional_probability)

print("\n================= BAYES THEOREM =================")

if len(categorical_cols) >= 2:

    # P(A)
    prob_A = len(
        df[df[event_A_col] == event_A]
    ) / total_rides

    # P(B)
    prob_B = len(
        df[df[event_B_col] == event_B]
    ) / total_rides

    # P(B|A)
    prob_B_given_A = intersection / len(
        df[df[event_A_col] == event_A]
    )

    # Bayes Theorem
    bayes_probability = (
        prob_B_given_A * prob_A
    ) / prob_B

    print(f"\nBayes Probability P({event_A}|{event_B}) =")
    print(bayes_probability)

print("\n=============== NORMAL DISTRIBUTION ===============")

if len(numeric_cols) > 0:

    numeric_col = numeric_cols[0]

    data = df[numeric_col]

    mean_value = np.mean(data)

    std_value = np.std(data)

    print(f"\nUsing Numeric Column: {numeric_col}")

    print(f"Mean: {mean_value}")

    print(f"Standard Deviation: {std_value}")

    # Probability for one value
    sample_value = mean_value

    normal_probability = norm.pdf(
        sample_value,
        mean_value,
        std_value
    )

    print(f"\nNormal Distribution Probability:")
    print(normal_probability)

print("\n============== BINOMIAL DISTRIBUTION ==============")

# Example assumptions
n = 10          # number of rides
p = 0.5         # success probability
x = 5           # successful rides

binomial_probability = binom.pmf(x, n, p)

print(f"\nBinomial Probability P(X={x}) =")
print(binomial_probability)
```

##Detect outliers using percentiles, quartiles, IQR, and z-score

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

url = "https://github.com/GeethaGunasekaran1/Dataset_rep/raw/refs/heads/main/Trip_Analysis.xlsx"

df = pd.read_excel(url)

numeric_df = df.select_dtypes(include=[np.number])

print("\nNumeric Columns:")
print(numeric_df.columns)

print("\n================ PERCENTILES ================\n")

for col in numeric_df.columns:
    p25 = np.percentile(numeric_df[col], 25)
    p50 = np.percentile(numeric_df[col], 50)
    p75 = np.percentile(numeric_df[col], 75)
    p95 = np.percentile(numeric_df[col], 95)

    print(f"\nColumn: {col}")
    print(f"25th Percentile : {p25:.2f}")
    print(f"50th Percentile : {p50:.2f}")
    print(f"75th Percentile : {p75:.2f}")
    print(f"95th Percentile : {p95:.2f}")

print("\n================ QUARTILES ================\n")

for col in numeric_df.columns:
    Q1 = numeric_df[col].quantile(0.25)
    Q2 = numeric_df[col].quantile(0.50)
    Q3 = numeric_df[col].quantile(0.75)

    print(f"\nColumn: {col}")
    print(f"Q1 : {Q1:.2f}")
    print(f"Q2 (Median) : {Q2:.2f}")
    print(f"Q3 : {Q3:.2f}")

print("\n================ IQR OUTLIERS ================\n")

iqr_outliers = {}

for col in numeric_df.columns:

    Q1 = numeric_df[col].quantile(0.25)
    Q3 = numeric_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = numeric_df[
        (numeric_df[col] < lower_bound) |
        (numeric_df[col] > upper_bound)
    ]

    iqr_outliers[col] = outliers

    print(f"\nColumn: {col}")
    print(f"IQR : {IQR:.2f}")
    print(f"Lower Bound : {lower_bound:.2f}")
    print(f"Upper Bound : {upper_bound:.2f}")
    print(f"Number of Outliers : {len(outliers)}")

print("\n================ Z-SCORE OUTLIERS ================\n")

zscore_outliers = {}

z_scores = np.abs(zscore(numeric_df))

for i, col in enumerate(numeric_df.columns):

    outliers = numeric_df[z_scores[:, i] > 3]

    zscore_outliers[col] = outliers

    print(f"\nColumn: {col}")
    print(f"Number of Outliers : {len(outliers)}")
```

#Apply Inferential Statistics (bootstrapping, CLT, point & interval estimation)

```python
target_column = numeric_cols[0]

data = df[target_column]

print("\nSelected Column for Inferential Statistics:")
print(target_column)

sample_size = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

print("\n" + "=" * 60)
print("SAMPLE INFORMATION")
print("=" * 60)

print(f"Sample Size       : {sample_size}")
print(f"Sample Mean       : {sample_mean:.4f}")
print(f"Sample Std Dev    : {sample_std:.4f}")

print("\n" + "=" * 60)
print("CENTRAL LIMIT THEOREM (CLT)")
print("=" * 60)

# Generate sampling distribution
sample_means = []

num_samples = 1000
sample_n = 30

for i in range(num_samples):

    sample = np.random.choice(data, size=sample_n, replace=True)

    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

clt_mean = np.mean(sample_means)
clt_std = np.std(sample_means)

print(f"Population Mean Approximation : {clt_mean:.4f}")
print(f"Sampling Distribution Std Dev : {clt_std:.4f}")

print("\nAccording to CLT:")
print("As sample size increases, sample means become normally distributed.")

print("\n" + "=" * 60)
print("POINT ESTIMATION")
print("=" * 60)

print(f"Point Estimate of Population Mean = {sample_mean:.4f}")

print("\n" + "=" * 60)
print("INTERVAL ESTIMATION (95% CONFIDENCE INTERVAL)")
print("=" * 60)

confidence_level = 0.95

# Standard Error
standard_error = sample_std / np.sqrt(sample_size)

# Z value for 95% confidence
z_value = stats.norm.ppf(0.975)

# Margin of Error
margin_error = z_value * standard_error

# Confidence Interval
lower_bound = sample_mean - margin_error
upper_bound = sample_mean + margin_error

print(f"Confidence Level : {confidence_level * 100}%")
print(f"Lower Bound      : {lower_bound:.4f}")
print(f"Upper Bound      : {upper_bound:.4f}")

print("\n" + "=" * 60)
print("BOOTSTRAPPING")
print("=" * 60)

bootstrap_means = []

bootstrap_iterations = 1000

for i in range(bootstrap_iterations):

    bootstrap_sample = np.random.choice(
        data,
        size=len(data),
        replace=True
    )

    bootstrap_means.append(np.mean(bootstrap_sample))

bootstrap_means = np.array(bootstrap_means)

bootstrap_mean = np.mean(bootstrap_means)

# Bootstrap Confidence Interval
bootstrap_ci_lower = np.percentile(bootstrap_means, 2.5)
bootstrap_ci_upper = np.percentile(bootstrap_means, 97.5)

print(f"Bootstrap Mean Estimate : {bootstrap_mean:.4f}")

print("\nBootstrap 95% Confidence Interval")
print(f"Lower Bound : {bootstrap_ci_lower:.4f}")
print(f"Upper Bound : {bootstrap_ci_upper:.4f}")

print("\n" + "=" * 60)
print("SUMMARY OF INFERENTIAL STATISTICS")
print("=" * 60)

print(f"""
1. Point Estimate:
   Estimated Population Mean = {sample_mean:.4f}

2. Interval Estimate:
   95% CI = ({lower_bound:.4f}, {upper_bound:.4f})

3. Bootstrapping:
   Bootstrap Mean = {bootstrap_mean:.4f}

4. Central Limit Theorem:
   Sampling distribution approximates normal distribution
   when sample size is sufficiently large.
""")
```

#●Conduct Hypothesis Testing (z-test, t-test, chi-square, F-test, ANOVA, A/B testing)

```python
col1 = numeric_cols[0]
col2 = numeric_cols[1]

data1 = df[col1]
data2 = df[col2]

print("\nSelected Columns:")
print("Column 1:", col1)
print("Column 2:", col2)
alpha = 0.05

print("\n" + "=" * 70)
print("1. Z-TEST")
print("=" * 70)

mean1 = np.mean(data1)
mean2 = np.mean(data2)

std1 = np.std(data1, ddof=1)
std2 = np.std(data2, ddof=1)

n1 = len(data1)
n2 = len(data2)

# Z Statistic
z_stat = (mean1 - mean2) / np.sqrt((std1**2 / n1) + (std2**2 / n2))

# P-value
p_value_z = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"Z Statistic : {z_stat:.4f}")
print(f"P-value     : {p_value_z:.4f}")

if p_value_z < alpha:
    print("Result      : Reject Null Hypothesis")
else:
    print("Result      : Fail to Reject Null Hypothesis")


print("\n" + "=" * 70)
print("2. INDEPENDENT T-TEST")
print("=" * 70)

t_stat, p_value_t = stats.ttest_ind(data1, data2)

print(f"T Statistic : {t_stat:.4f}")
print(f"P-value     : {p_value_t:.4f}")

if p_value_t < alpha:
    print("Result      : Reject Null Hypothesis")
else:
    print("Result      : Fail to Reject Null Hypothesis")

print("\n" + "=" * 70)
print("3. CHI-SQUARE TEST")
print("=" * 70)

# Create categorical bins
cat1 = pd.cut(data1, bins=5)
cat2 = pd.cut(data2, bins=5)

# Contingency Table
contingency_table = pd.crosstab(cat1, cat2)

chi2_stat, p_value_chi, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic : {chi2_stat:.4f}")
print(f"P-value              : {p_value_chi:.4f}")
print(f"Degrees of Freedom   : {dof}")

if p_value_chi < alpha:
    print("Result               : Variables are dependent")
else:
    print("Result               : Variables are independent")

print("\n" + "=" * 70)
print("4. F-TEST")
print("=" * 70)

variance1 = np.var(data1, ddof=1)
variance2 = np.var(data2, ddof=1)

f_stat = variance1 / variance2

df1 = n1 - 1
df2 = n2 - 1

p_value_f = 1 - stats.f.cdf(f_stat, df1, df2)

print(f"F Statistic : {f_stat:.4f}")
print(f"P-value     : {p_value_f:.4f}")

if p_value_f < alpha:
    print("Result      : Variances are significantly different")
else:
    print("Result      : Variances are similar")

print("\n" + "=" * 70)
print("5. ANOVA TEST")
print("=" * 70)

# Create 3 sample groups from data1
group1 = data1.sample(frac=0.33, random_state=1)
group2 = data1.sample(frac=0.33, random_state=2)
group3 = data1.sample(frac=0.33, random_state=3)

anova_stat, p_value_anova = stats.f_oneway(group1, group2, group3)

print(f"ANOVA Statistic : {anova_stat:.4f}")
print(f"P-value         : {p_value_anova:.4f}")

if p_value_anova < alpha:
    print("Result          : At least one group mean differs")
else:
    print("Result          : Group means are similar")

print("\n" + "=" * 70)
print("6. A/B TESTING")
print("=" * 70)

# Simulated A/B Groups
A_group = data1.sample(frac=0.5, random_state=10)
B_group = data1.sample(frac=0.5, random_state=20)

ab_t_stat, ab_p_value = stats.ttest_ind(A_group, B_group)

print(f"A/B Test Statistic : {ab_t_stat:.4f}")
print(f"P-value            : {ab_p_value:.4f}")

if ab_p_value < alpha:
    print("Result             : Significant difference between A & B")
else:
    print("Result             : No significant difference between A & B")
```
